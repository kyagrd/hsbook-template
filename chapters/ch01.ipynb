```haskell
-- 정수 연산
let x = 42 :: Int
let y = 17 :: Int
print (x + y)
print (x - y)
print (x * y)
print (div x y)
```

```haskell
-- 부동소수점 연산
let pi' = 3.14159265 :: Double
let r   = 5.0 :: Double
print (pi' * r * r)
```

Haskell에서 함수는 일급 시민(first-class citizen)입니다.

```haskell
-- 간단한 함수 정의
square :: Int -> Int
square n = n * n

square 7
```

```haskell
-- 조건식 (if-then-else)
absolute :: Int -> Int
absolute n = if n < 0 then -n else n

absolute (-5)
```

## 타입(Type)

Haskell은 강력한 정적 타입 시스템을 갖추고 있습니다.
`:type` 명령은 **IHaskell REPL 전용 명령**으로, Jupyter Notebook에서 표현식의 타입을 확인할 때 사용합니다.
(일반 `.hs` 소스 파일에서는 사용할 수 없으며, GHCi 인터랙티브 환경에서도 동일하게 동작합니다.)

```haskell
:type True
:type 'a'
:type "hello"
:type 42
:type 3.14
```

## 리스트(List)

Haskell의 리스트는 동일 타입의 원소를 순서 있게 담는 자료구조입니다.

```haskell
let nums = [1, 2, 3, 4, 5] :: [Int]
print nums
print (head nums)
print (tail nums)
print (length nums)
```

```haskell
-- 리스트 덧붙이기
let a = [1, 2, 3] :: [Int]
let b = [4, 5, 6] :: [Int]
print (a ++ b)
```

```haskell
-- 범위 표현
print [1..10]
print [2, 4..20]
```

```haskell
-- 오류 발생
1 / 0          -- 이건 무한대
1 / (0 :: Int) -- 이건 에러
```
